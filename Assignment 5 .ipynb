```python
class Bank_Account:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            print(f"Cash Deposite {amount}")
        else:
            print('cash cant deposite')

    def with_draw(self,amount):
        if amount <= 0:
            print("Cash not available")
        elif amount < 0:
            print("Transaction fail")
        else:
            self.__balance -= amount
            print("Amount Withdraw successfully")

    def get_balance(self):
        return self.__balance
        
```

```python
ba=Bank_Account(5000)
```

```python
ba.deposit(4000)
```

```python
ba.get_balance()
```

```python
ba.with_draw(2000)
```

```python
ba.get_balance()
```

# Answer No 17

```python
from abc import ABC, abstractmethod
```

```python
class Computer(ABC):
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    #@abstractmethod
    #def process(self):
     #   pass

class Laptop(Computer):
    def __init__(self, brand, model, processor):
        super().__init__(brand, model)
        self.processor = processor

    def process(self):
        return f"""
        Brand: {self.brand}
        Model: {self.model}
        Processor: {self.processor}"""


class Desktop(Computer):
    def __init__(self, brand, model, monitor):
        super().__init__(brand, model)
        self.monitor = monitor

    def process(self):
        return f"""
        Brand: {self.brand}
        Model: {self.model}
        Monitor: {self.monitor}"""
```

```python
my_comp = Laptop('DELL','core i7', 'Core i7')
print(my_comp.process())
```

```python
my_desktop = Desktop('HP', 'Pavilion', 'Samsung')
print(my_desktop.process())
```

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):
    def start_engine(self):
        print("Engine started!")

    
    @abstractmethod
    def drive(self):
        pass

class Honda(Car):
    def drive(self):
        print("Car is driving!")
```

```python
car = Honda()
```

```python
car.start_engine()
```

```python
car.drive()
```

# Answer No 20


```python

from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):

    def pay(self, amount):
        print(f"Paid Rs. {amount} using Credit Card")

class UPIPayment(Payment):

    def pay(self, amount):
        print(f"Paid Rs. {amount} using UPI")

```

```python
p1 = CreditCardPayment()
p1.pay(5000)
```

```python
p2 = UPIPayment()
p2.pay(2000)
```
